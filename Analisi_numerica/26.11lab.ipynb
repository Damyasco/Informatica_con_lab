#### Perché F(x) è strettamente crescente?
$F(x)= k \cdot x +\alfa \cdot x^3$ la sua derivata $F(x)' = k + 2\alfa x^2$ che per $\alfa$, k, positive è stremente positiva,
quindi F è strettamente screscente, è anche suriettica e quindi ammette un'inversa.
Calcolo F(x) per 5, 9 e 17 allungamenti 

```python
import numpy as np
import matplotlib.pyplot as plt
k, x_max ,alfa = 10, 1, 1

def allungamenti(x_m, n):
    return np.linspace(0, x_m, n)

def f(x):
    return k*x + alfa * x**3

npt = 17
x = allungamenti(x_max, npt)
F = f(x)

V = np.vander(F)
c = np.linalg.solve(V,x)

x_controllo  = np.linspace(0, x_max, 200)
f_controllo = f(x_controllo)

plt.plot(f_controllo, np.polyval(c, f_controllo))
plt.plot(F,x, 'o')
plt.xlabel('Forza')
plt.ylabel('Allungamento')

plt.figure()
residui= x_controllo - np.polyval(c, f_controllo)
plt.plot(f_controllo, residui, '.', label='Residui')
plt.axhline(0, color='black')
plt.xlabel('Forza')
plt.ylabel('Residui')
plt.show()
# ridefinisco la forza in modo da poter variare alfa
def f(x, alfa):
    return k*x + alfa * x**3

# residui massimi
res_max = []
alfas_s  = [1, 5 ,10]
for i in range(3):
    F = f(x, alfas_s[i])
    V = np.vander(F)
    c = np.linalg.solve(V,x)
    x_controllo  = np.linspace(0, x_max, 200)
    f_controllo = f(x_controllo, alfas_s[i])
    residui= x_controllo - np.polyval(c, f_controllo)
    res_max.append(np.max(np.abs(residui)))

print(res_max)
    
```
